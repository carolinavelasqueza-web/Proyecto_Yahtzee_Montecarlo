#Actividad: Simulación del juego Yahtzee mediante el método de Montecarlo


Punto 1. Importar las herramientas necesarias

```python
# Herramientas para generar dados, organizar resultados y calcular estadísticas.
import random
import statistics
import matplotlib.pyplot as plt
from collections import Counter
```

Punto 2. Definir las categorías de puntuación

```python
# Lista con las trece categorías del juego Yahtzee.
CATEGORIAS = [
    "Unos",
    "Doses",
    "Treses",
    "Cuatros",
    "Cincos",
    "Seises",
    "Trio",
    "Poker",
    "Full",
    "Escalera pequeña",
    "Escalera grande",
    "Chance",
    "Yahtzee"
]

# Mostramos las categorías para comprobar que la lista está completa.
print("Categorías del juego:")
for numero, categoria in enumerate(CATEGORIAS, start=1):
    print(f"{numero}. {categoria}")

print("\nTotal de categorías:", len(CATEGORIAS))
```

Punto 3. Simular el lanzamiento de los dados

```python
def lanzar_dados(cantidad):
    """
    Simula el lanzamiento de una cantidad determinada de dados.

    Parámetro:
        cantidad: número de dados que se van a lanzar.

    Retorna:
        Una lista con los resultados aleatorios entre 1 y 6.
    """
    return [random.randint(1, 6) for _ in range(cantidad)]


# Probamos la función lanzando los cinco dados.
dados_prueba = lanzar_dados(5)

print("Resultados de los cinco dados:", dados_prueba)
print("Cantidad de dados:", len(dados_prueba))
```

Punto 4. Contar las caras obtenidas

```python
def contar_caras(dados):
    """
    Cuenta cuántas veces aparece cada cara en los dados.

    Parámetro:
        dados: lista con los resultados obtenidos.

    Retorna:
        Un contador con la frecuencia de cada número.
    """
    return Counter(dados)


# Contamos las caras de los dados de prueba.
frecuencias = contar_caras(dados_prueba)

print("Dados obtenidos:", dados_prueba)
print("Frecuencia de cada cara:", dict(frecuencias))
```

Punto 5. Calcular los puntos de cada combinación

```python
def puntuar(dados, categoria):
    """
    Calcula los puntos de los dados para una categoría de Yahtzee.

    Parámetros:
        dados: lista con los cinco resultados.
        categoria: categoría en la que se quieren calcular los puntos.

    Retorna:
        La puntuación obtenida en esa categoría.
    """

    # Contamos las repeticiones y sumamos el valor de los dados.
    conteo = contar_caras(dados)
    total = sum(dados)
    valores = set(dados)

    # Categorías numéricas: se suman las caras iguales al número elegido.
    if categoria in CATEGORIAS[:6]:
        cara = CATEGORIAS.index(categoria) + 1
        return dados.count(cara) * cara

    # Trío: tres o más dados con el mismo número.
    if categoria == "Trio":
        return total if max(conteo.values()) >= 3 else 0

    # Póquer: cuatro o más dados con el mismo número.
    if categoria == "Poker":
        return total if max(conteo.values()) >= 4 else 0

    # Full: tres dados iguales y otros dos iguales.
    if categoria == "Full":
        return 25 if sorted(conteo.values()) == [2, 3] else 0

    # Escalera pequeña: cuatro números consecutivos como mínimo.
    if categoria == "Escalera pequeña":
        secuencias = [
            {1, 2, 3, 4},
            {2, 3, 4, 5},
            {3, 4, 5, 6}
        ]

        return 30 if any(
            secuencia.issubset(valores)
            for secuencia in secuencias
        ) else 0

    # Escalera grande: cinco números consecutivos.
    if categoria == "Escalera grande":
        return 40 if valores in (
            {1, 2, 3, 4, 5},
            {2, 3, 4, 5, 6}
        ) and len(valores) == 5 else 0

    # Chance: se suman los cinco dados, sin exigir una combinación.
    if categoria == "Chance":
        return total

    # Yahtzee: los cinco dados deben mostrar el mismo número.
    if categoria == "Yahtzee":
        return 50 if max(conteo.values()) == 5 else 0

    # Si se recibe una categoría no reconocida, no se asignan puntos.
    return 0
```

```python
# Probamos diferentes combinaciones de dados.
pruebas = [
    ([2, 2, 2, 5, 6], "Doses"),
    ([3, 3, 3, 4, 5], "Trio"),
    ([4, 4, 4, 4, 2], "Poker"),
    ([2, 2, 3, 3, 3], "Full"),
    ([1, 2, 3, 4, 6], "Escalera pequeña"),
    ([2, 3, 4, 5, 6], "Escalera grande"),
    ([6, 6, 6, 6, 6], "Yahtzee"),
    ([1, 2, 3, 4, 5], "Chance")
]

# Mostramos los puntos obtenidos en cada prueba.
for dados, categoria in pruebas:
    puntos = puntuar(dados, categoria)
    print(f"Dados: {dados} | Categoría: {categoria} | Puntos: {puntos}")
```

Punto 6. Elegir qué dados conservar

```python
def elegir_dados_a_conservar(dados):
    """
    Selecciona los dados que se conservarán para el siguiente lanzamiento.

    Se conserva la cara más repetida. En caso de empate,
    se elige la cara de mayor valor.
    """
    # Identificamos las frecuencias de las caras.
    conteo = contar_caras(dados)

    # Elegimos la cara con mayor frecuencia y, en caso de empate,
    # la de mayor valor.
    cara_objetivo = max(
        conteo,
        key=lambda cara: (conteo[cara], cara)
    )

    # Devolvemos únicamente los dados que se conservarán.
    conservados = [
        dado for dado in dados
        if dado == cara_objetivo
    ]

    return conservados


# Probamos la estrategia con una combinación conocida.
dados_prueba = [4, 4, 4, 2, 6]
conservados = elegir_dados_a_conservar(dados_prueba)

print("Dados iniciales:", dados_prueba)
print("Dados que se conservan:", conservados)
print("Dados que se volverían a lanzar:", 5 - len(conservados))
```

Punto 7. Simular un turno completo

```python
def jugar_turno():
    """
    Simula un turno con un máximo de tres lanzamientos.

    Retorna:
        Los cinco dados que quedan al finalizar el turno.
    """

    # Primer lanzamiento: se lanzan los cinco dados.
    dados = lanzar_dados(5)
    print("Primer lanzamiento:", dados)

    # Se realizan como máximo dos lanzamientos adicionales.
    for numero_lanzamiento in range(2):

        # Elegimos los dados que queremos conservar.
        conservados = elegir_dados_a_conservar(dados)

        # Calculamos cuántos dados necesitamos volver a lanzar.
        cantidad_nuevos = 5 - len(conservados)

        # Si los cinco dados están conservados, termina el turno.
        if cantidad_nuevos == 0:
            print("Se conservan los cinco dados. Termina el turno.")
            break

        # Lanzamos únicamente los dados que no conservamos.
        nuevos = lanzar_dados(cantidad_nuevos)

        # Combinamos los dados conservados con los nuevos resultados.
        dados = conservados + nuevos

        print(
            f"Lanzamiento {numero_lanzamiento + 2}:",
            dados
        )

    return dados


# Ejecutamos una prueba de un turno.
dados_finales = jugar_turno()

print("\nDados finales del turno:", dados_finales)
```

Punto 8. Simular los trece turnos de un jugador

```python
def jugar_jugador():
    """
    Simula los trece turnos de un jugador.

    Retorna:
        La puntuación total y el detalle de puntos por categoría.
    """

    # Cada jugador comienza con todas las categorías disponibles.
    categorias_disponibles = CATEGORIAS.copy()

    # Aquí guardaremos los puntos obtenidos en cada categoría.
    puntuaciones = {}

    # Repetimos el proceso una vez por cada categoría.
    for turno in range(13):

        print(f"\nTurno {turno + 1}")

        # Simulamos los lanzamientos del turno.
        dados = jugar_turno()

        # Calculamos los puntos posibles en cada categoría disponible.
        opciones = [
            (puntuar(dados, categoria), categoria)
            for categoria in categorias_disponibles
        ]

        # Elegimos la categoría con mayor puntuación.
        puntos, categoria_elegida = max(
            opciones,
            key=lambda opcion: opcion[0]
        )

        # Registramos los puntos y retiramos la categoría utilizada.
        puntuaciones[categoria_elegida] = puntos
        categorias_disponibles.remove(categoria_elegida)

        print("Categoría elegida:", categoria_elegida)
        print("Puntos obtenidos:", puntos)

    # Sumamos las seis categorías numéricas superiores.
    puntos_superiores = sum(
        puntuaciones[categoria]
        for categoria in CATEGORIAS[:6]
    )

    # Se otorgan 35 puntos adicionales si se alcanzan 63 puntos.
    bono = 35 if puntos_superiores >= 63 else 0

    # Calculamos la puntuación final.
    total = sum(puntuaciones.values()) + bono

    print("\nPuntos de las categorías superiores:", puntos_superiores)
    print("Bono superior:", bono)
    print("Puntuación total del jugador:", total)

    return total, puntuaciones


# Probamos una partida individual de un jugador.
puntuacion, detalle = jugar_jugador()
```
